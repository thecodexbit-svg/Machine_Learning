# PANDAS

**Pandas** is a popular Python library used for **data manipulation and data analysis**. It makes it easy to work with structured data such as tables, CSV files, and Excel files.

## Main Data Structures

* **Series** — A one-dimensional labeled data structure, similar to a single column.
* **DataFrame** — A two-dimensional table with rows and columns, similar to an Excel spreadsheet.

## What Can Pandas Do?

* Read and write CSV and Excel files
* Filter and sort data
* Handle missing values
* Add, remove, and modify columns
* Group and summarize data
* Perform data analysis and statistics

## Example

```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya"],
    "Age": [20, 21, 19]
}

df = pd.DataFrame(data)

print(df)
```

### Output

```text
    Name  Age
0   Amit   20
1  Rahul   21
2  Priya   19
```

> **In short:** Pandas is like a powerful programming version of Excel tables, mainly used for working with and analyzing data in Python.


````python
import pandas as pd
import numpy as np
````

````python
# pd.DataFrame() - creates a table from dictionary
data = {
    'table_no' : [12,13,14,15,16],
    'bill_amount' : [100,10000,100,0,10],
    'waiter_name' : ['Anugrah', 'Sharief', 'Preetam' , 'Pratham', 'Ayush']
}
df = pd.DataFrame(data)

print(df)
print("Data Type :",type(df))

print(df['table_no'])
print(type(df['table_no']))
````

````python
# .values / .to_numpy() - the NumPy array underneath
print(df.values)
print(type(df.values))
print()
print(df.to_numpy())
print(type(df.to_numpy()))
````

````python
# pd.Series() - 1D data with an index
total_bill = [4700,5678,1253,1000]
s = pd.Series(total_bill, index = ['mon','tue','wed','fri'])
print(s)
print(type(s))
````

````python
# Q1 --
s = [45,30,15,60]
s = pd.Series(s ,index = ['PIZZA','PASTA','FRIES','BURGER'])
print(s)
````

````python
# Q2 --
p = {
    'Brand' : ['Puma', 'Campus','Nike'],
    'Price' : [2500,6000,7900]
}
df = pd.DataFrame(p)

print(df)
````
